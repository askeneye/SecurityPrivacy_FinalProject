Single out surveyed people in public dataset

```python
import pandas as pd
from pathlib import Path
# Path to data folder
DATA_DIR = Path("../data_V/")
with open(DATA_DIR / "survey_listV.txt", "r", encoding="utf-8") as file:
    names = file.read().splitlines()

public_data_all= pd.read_excel(DATA_DIR / "public_data_registerV.xlsx")
public_df = public_data_all[public_data_all["name"].isin(names)]
print(len(public_df))
```

Applying the same anonymisation technique as V group

```python
anonym_df = pd.read_csv(DATA_DIR / "anonymised_dataV.csv")

reference_date=pd.Timestamp("2025-07-01")  #Date of conducting the interview

dob = pd.to_datetime(public_df["dob"])
public_df["age"]=  (
    reference_date.year - dob.dt.year
    - (
        (dob.dt.month > reference_date.month)
        | (
            (dob.dt.month == reference_date.month)
            & (dob.dt.day > reference_date.day)
        )
    ).astype(int)
)
public_df["age"] = pd.cut(
    public_df["age"],
    bins=[20, 29, 39, 49, 59,69, 120],
    labels=["20-29", "30-39", "40-49", "50-59", "60-69", "70+"]
)
public_df.drop(columns=["dob"], inplace=True)
public_df.head()
```

```python
public_df["marital_status"] = public_df["marital_status"].replace({
    "Married/separated": "Married/Separated",
    "Divorced": "Not Married",
    "Widowed": "Not Married",
    "Never married": "Not Married"
})
print(public_df["marital_status"].value_counts())
print(anonym_df["marital_status"].value_counts())
```

```python
qi = ["sex","marital_status","evote","zip","age"]
for column in qi:
    print(column)
    print("Anonymous:")
    print(anonym_df[column].value_counts())

    print("Public:")
    print(public_df[column].value_counts())
```

```python
group_sizes = anonym_df.groupby(qi, observed=True).size()
group_sizes_2 = public_df.groupby(qi, observed=True).size()

comparison = pd.concat(
    [group_sizes, group_sizes_2],
    axis=1,
    keys=["Anonymous", "Public"]
)

comparison.head()
```
